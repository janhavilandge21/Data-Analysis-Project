# AI SQL Generator

```python
from langchain_groq import ChatGroq

# Initialize LLM
llm = ChatGroq(
    groq_api_key="",
    model_name="llama3-70b-8192"
)

def generate_sql(question):

    prompt = f"""
    Convert the following question into SQL query.

    Table name: sales

    Columns:
    Order_ID
    Region
    Category
    Sales
    Profit
    Quantity

    Question: {question}

    SQL:
    """

    response = llm.invoke(prompt)

    return response.content
```
